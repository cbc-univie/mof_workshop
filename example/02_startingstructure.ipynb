### A first look

We’ll start by loading the selected MOF and inspecting its cell, composition, and basic geometry.
For quick inspections you can use:

```python
from ase.io import read
from ase.visualize import view
```

### Loading Coordinates

We can load the coordinates into an ASE Atoms object. You can name the variable whatever you like; most examples use atoms.

```python
atoms = read("02_uio-66.cif")
```

### Visualizing structures and trajectories

ASE includes an integrated viewer for quick visualization of structures and, later, trajectories; if you prefer external tools, you can also use VMD/Avogadro/Mercury/Ovito to view your coordinates.

```python
view(atoms)
```

### Basic structure info

Let’s print the number of atoms, the unit cell (a, b, c, α, β, γ), and the cell volume.

```python
len(atoms)
```

```python
atoms.get_volume()
```

```python
atoms.cell.cellpar()
```

### Now we add our adsorbate 

You can choose either CO2 or H2O.
We need the following module for that:

```python
from ase.build import molecule
```

```python
co2 = molecule('CO2')
#h2o = molecule('H2O')
```

We place the molecule in the middle of the cell as the framework has no atoms there

```python
com = atoms.get_cell().sum(axis=0) / 2
co2.translate(com )
```

Now we simply combine the two structures

```python
system = atoms + co2
```

```python
system.wrap()
```

Quick sanity check? 

We can verify the chemical formula or the length of the atoms object

```python
atoms.get_chemical_formula()
```

```python
system.get_chemical_formula()
```

We see that we added one C and two O. 

```python
len(atoms)
```

```python
len(system)
```

We can see that we have three atoms more.

We can also take a look at the system.

```python
view(system)
```

### Save the new combined structure

Now that we evaluated the structure is correct we can save it to an input structure. ASE can write many coordinate formats such as cif/PDB or xyz.

```python
from ase.io import write
```

```python
write("03_structure.cif", system)
```
