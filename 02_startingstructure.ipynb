A first look

First we want to take a look at our selected Metal-Organic Framework
We can easily view structures and later trajectories with ase.visualize.view

We import the modules we need

```python
from ase.io import read
from ase.visualize import view
```

We can load the coordinates as an atoms object. You can call it however you want, in most cases online you will find it called "atoms"

```python
atoms = read("02_uio-66.cif")
```

There is an integrated viewer which makes it very easy to visualize your coordinates and even later on trajectories. If you have other programs you prefer, you can use for example vmd to visualize the coordinates too.

```python
view(atoms)
```
